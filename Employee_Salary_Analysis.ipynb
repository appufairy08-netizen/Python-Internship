```python
salaries=[20000,23000,15000,40000,52000]
total_num_salaries=len(salaries)
print("Number of Employees:",total_num_salaries)
total_salary=sum(salaries)
print("Total Salary:",total_salary)
average_salaries=(total_salary/total_num_salaries)
print("Average Salary:",average_salaries)
highest_salary=max(salaries)
print("Highest Salary:",highest_salary)
lowest_salary=min(salaries)
print("Lowest Salary:",lowest_salary)
```
